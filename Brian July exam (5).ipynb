```python
num = int(input("Enter number: "))
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num //= 10

print("Reversed Number:", reverse)


num = int(input("Enter number: "))
original = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num //= 10

if original == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")

     
```

```python
students = {}

while True:
    print("\n1. Add Student")
    print("2. Search Student")
    print("3. Delete Student")
    print("4. Display Students")
    print("5. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        roll = input("Roll Number: ")
        name = input("Student Name: ")
        students[roll] = name

    elif choice == "2":
        roll = input("Enter Roll Number: ")
        print(students.get(roll, "Student Not Found"))

    elif choice == "3":
        roll = input("Enter Roll Number: ")
        if roll in students:
            del students[roll]
            print("Deleted Successfully")
        else:
            print("Student Not Found")

    elif choice == "4":
        for roll, name in students.items():
            print(roll, ":", name)

    elif choice == "5":
        break

    else:
        print("Invalid Choice")

```

```python
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
op = input("Enter operator (+,-,*,/): ")

if op == "+":
    print("Result =", a + b)
elif op == "-":
    print("Result =", a - b)
elif op == "*":
    print("Result =", a * b)
elif op == "/":
    if b == 0:
        print("Cannot divide by zero")
    else:
        print("Result =", a / b)
else:
    print("Invalid operator") 
```

```python
#Question 10 employee menu with classess
```

```python
class Employee:

    def __init__(self, empid, name, dept, salary):
        self.empid = empid
        self.name = name
        self.dept = dept
        self.salary = salary

employees = {}

while True:
    print("\n1.Add")
    print("2.Display")
    print("3.Search")
    print("4.Update Salary")
    print("5.Delete")
    print("6.Exit")

    ch = input("Choice: ")

    try:
        if ch == "1":
            empid = input("Employee ID: ")
            name = input("Name: ")
            dept = input("Department: ")
            salary = float(input("Salary: "))
            employees[empid] = Employee(empid, name, dept, salary)

        elif ch == "2":
            for e in employees.values():
                print(e.empid, e.name, e.dept, e.salary)

        elif ch == "3":
            empid = input("Employee ID: ")
            if empid in employees:
                e = employees[empid]
                print(e.empid, e.name, e.dept, e.salary)
            else:
                print("Not Found")

        elif ch == "4":
            empid = input("Employee ID: ")
            employees[empid].salary = float(input("New Salary: "))

        elif ch == "5":
            empid = input("Employee ID: ")
            del employees[empid]

        elif ch == "6":
            break

        else:
            print("Invalid Choice")

    except Exception as e:
        print("Error:", e)
   
```

```python
#Question 12 inventory management
```

```python
inventory = {}

filename = "inventory.txt"

while True:

    print("\n1.Add Product")
    print("2.Search Product")
    print("3.Update Quantity")
    print("4.Delete Product")
    print("5.Save to File")
    print("6.Read from File")
    print("7.Exit")

    choice = input("Choice: ")

    try:
        if choice == "1":
            pid = input("Product ID: ")
            name = input("Product Name: ")
            price = float(input("Price: "))
            qty = int(input("Quantity: "))

            inventory[pid] = [name, price, qty]

        elif choice == "2":
            pid = input("Product ID: ")
            print(inventory.get(pid, "Product Not Found"))

        elif choice == "3":
            pid = input("Product ID: ")
            inventory[pid][2] = int(input("New Quantity: "))

        elif choice == "4":
            pid = input("Product ID: ")
            del inventory[pid]

        elif choice == "5":
            with open(filename, "w") as file:
                for pid, data in inventory.items():
                    file.write(f"{pid},{data[0]},{data[1]},{data[2]}\n")
            print("Saved Successfully")

        elif choice == "6":
            with open(filename, "r") as file:
                print(file.read())

        elif choice == "7":
            break

        else:
            print("Invalid Choice")

    except Exception as e:
        print("Error:", e)
```
